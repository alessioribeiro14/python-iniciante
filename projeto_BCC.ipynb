```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

x = np.arange(0,500, 0.1)
y1 = 4*x**2 + x
y2 = 10*x + 4
plt.plot(x,y1, 'b.-', label='Y1')
plt.plot(x,y2, 'r.', label='Y2')
plt.grid('y')
plt.xlabel('Dominio')
plt.ylabel('Contradominio')
plt.title('Funções')
plt.legend()
```

```python
df = pd.DataFrame(y1, y2)
df
```

```python
c = np.corrcoef(y1, y2)
c
```

Existe correlação linear positiva, e proxima de 1, ou seja muito forte.
